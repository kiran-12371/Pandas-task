```python
#18. Combinig data frames
df1 = {
    "Platform" : ["Instagram","TikTok","Reddit","Youtube"],
    "Daily_usage_hours" : [4,3,1,2]
}

df2 = {
    "Platform" : ["Linkedin","Snapchat","Facebook","Whatsapp","Discord"],
    "Daily_usage_hours" : [3,2,1,5,3]
}

df1 = pd.DataFrame(df1)
df2 = pd.DataFrame(df2)
print("=====DF1======")
print(df1)
print("======DF======")
print(df2)
```

```python
final_result=pd.concat([df1,df2],ignore_index=True)
print(final_result)
```

```python
students={"ID":[1,2,3],"Name":["Ali","Sara","Ahmed"]}
marks={"ID":[1,2,3],"Marks":[70,85,60]}
df1=pd.DataFrame(students)
df2=pd.DataFrame(marks)
print(df1)
print(df2)
```

```python
#merge
final_result=pd.merge(df1,df2,on="ID")
print(final_result)
```
